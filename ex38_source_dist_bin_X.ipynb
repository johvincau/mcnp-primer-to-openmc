Example 38 shows the binning of particles based on which source distribution was sampled with
```
sdef pos=d1 erg=fpos d2
si1 L 5 3 6 75 3 6
sp1 0.3 0.7
ds2 S 3 4
si3 H 2 10 14
sp3 D 0 1 2
si4 H 5 2 8
sp4 D 0 3 1
f2:n 2
ft2 scd
fu2 3 4
```
which is a neutron flux tally on cell 2 modified by `ft2 scd` to bin against the source term `sdef pos=d1 erg=fpos d2`

OpenMC does not appear to have provisions for filtering based on source distribution sampling specifically. The source distributions would need to occupy distinct and non-overlapping regions of phase space in order to be able to specify tally filters that entirely encapsulate either one or the other.
