```python
import openmc
import numpy as np
import math
```

```python
# Material specification
m1 = openmc.Material()
p = 'ao'
m1.add_nuclide('C12',0.000125,p)
m1.add_nuclide('N14',0.6869,p)
m1.add_nuclide('O16',0.30124,p)
m1.add_nuclide('Ar40',0.011717,p)
m1.set_density('g/cc',0.001124)

m2 = openmc.Material()
m2.add_element('Pb',1,p)
m2.set_density('g/cc',11)

materials = openmc.Materials([m1,m2])
```

```python
# Geometry specification
s11 = openmc.Sphere(r=30)
s21 = openmc.Sphere(r=32,boundary_type='vacuum')

c1 = openmc.Cell(fill=m1,region=-s11)
c2 = openmc.Cell(fill=m2,region=+s11&-s21)
c3 = openmc.Cell(fill=None,region=+21)

geometry = openmc.Geometry([c1,c2,c3])
```

```python
# Source specification
source = openmc.IndependentSource()
source.spatial = openmc.stats.Point((0,0,0))
source.energy = openmc.stats.Delta(2.5*10**6)

settings = openmc.Settings()
settings.run_mode = 'fixed source'
settings.particles = 100
settings.batches = 100
settings.source = source
```
