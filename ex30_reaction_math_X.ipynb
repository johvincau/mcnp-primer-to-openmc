If we have 3 reactions, r1, r2, and r3, and we want to add r2 and r3, then multiply the result by r1..
then in OpenMC, it would make more sense to simply create a tally that scores r1, r2, and r3.

Then after running, manipulate the tally dataframe to yield the mean and std. dev. of each reaction rate and perhaps use the `uncertainties` module to carry uncertainty during the computation of (r2+r3) * r1. 

The MCNP primer states one should do `FMn (C m (r_1 r_2:r_1 r_3))`, since multiplication occurs first (basically one needs to distribute the multiplier r_1 to both r_2 and r_3 independently).
