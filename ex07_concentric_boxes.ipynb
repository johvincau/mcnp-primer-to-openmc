```python
%run -i 'clean_script.py'
import openmc
model = openmc.model.Model()
```

```python
# Using "macrobodies"
side = 2
s1 = openmc.model.RectangularParallelepiped(-side,side,-side,side,-side,side)
side += 2
s2 = openmc.model.RectangularParallelepiped(-side,side,-side,side,-side,side)
side += 2
s3 = openmc.model.RectangularParallelepiped(-side,side,-side,side,-side,side)
c1 = openmc.Cell(region=-s1)
c2 = openmc.Cell(region=+s1&-s2)
c3 = openmc.Cell(region=+s2&-s3)
c4 = openmc.Cell(region=+s3)

geometry = openmc.Geometry([c1,c2,c3,c4])
model.geometry = geometry
geometry.plot(width=(20,20),color_by='cell',outline=True)
```
