Example 25 describes how to use and save a structured mesh on which MCNP can do transport. In OpenMC however, there is only support for surface tracking, and not through a finite element mesh. While unstructured meshes as provided by optional OpenMC dependency`libmesh` can be used for binning tallies (as is relevant in coupled-code multiphysics problems, wherein the UM elements need to be kept consistent across codes), they are not for transport.
