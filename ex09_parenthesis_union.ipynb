Showcases the importance of parenthesis operators when unioning regions.

```python
import openmc as omc

# Surface definitions
l1 = 2
s1 = omc.model.RectangularParallelepiped(-l1, l1, -l1, l1, -l1, l1)
s2 = omc.Sphere(r=5)

s3= omc.Sphere(x0=15, r=5)
s4 = omc.Sphere(x0=21,r=8)
```

&: intersection, logical AND, (MCNP's ' ')
|: union, logical OR, (MCNP's ':')
~: complement, logical NOT, (MCNP's '#')

```python
# Cell definitions
c9 = omc.Cell(fill=None,region=-s1)
c17 = omc.Cell(fill=None,region= (-s2 & +s1) | (-s4 | -s3))
c22 = omc.Cell(fill=None,region= (+s2 & (+s3 | +s4)))
geometry=omc.Geometry([c9,c17,c22])
geometry.export_to_xml()

material=omc.Materials()
material.export_to_xml()

setting = omc.Settings()
setting.export_to_xml()
```
