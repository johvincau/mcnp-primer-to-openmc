Showcases the importance of parenthesis operators when unioning regions.

```python
import openmc 

# Surface definitions
l1 = 2
s1 = openmc.model.RectangularParallelepiped(-l1, l1, -l1, l1, -l1, l1)
s2 = openmc.Sphere(r=7)

s3= openmc.Sphere(x0=20, r=5)
s4 = openmc.Sphere(x0=25,y0=5,r=8)
```

```python
# Cell definitions
c9 = openmc.Cell(fill=None,region=-s1)
c17 = openmc.Cell(fill=None,region= (-s2 & +s1) | (-s4 | -s3))
c22 = openmc.Cell(fill=None,region= (+s2 & (+s3 | +s4)))
geometry=openmc.Geometry([c9,c17,c22])
geometry.plot(origin=(13,0,0),width=(45,30),color_by='cell',outline=True)
geometry.export_to_xml()

material=openmc.Materials()
material.export_to_xml()

setting = openmc.Settings()
setting.export_to_xml()
```
