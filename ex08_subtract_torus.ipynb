This example showcases the use of a torus to subtract from the inner sphere.

```python
import openmc 

# Surface definitions
s1 = openmc.Sphere(r=5)
s2 = openmc.Sphere(r=10)
s3 = openmc.XTorus(x0=4,y0=0,z0=0,a=4,b=1.5,c=1.5)
```

&: intersection, logical AND, (MCNP's ' ')
|: union, logical OR, (MCNP's ':')
~: complement, logical NOT, (MCNP's '#')

```python
# Cell definitions
c1 = openmc.Cell(cell_id=1,fill=None,region=-s1 & +s3)
c2 = openmc.Cell(cell_id=2,fill=None,region= -s2 | (+s1 & -s3))
c3 = openmc.Cell(cell_id=3,fill=None,region=+s2)

geometry=openmc.Geometry([c1,c2,c3])
geometry.plot(origin=(0,0,0),width=(20,20),color_by='cell',outline=True)

geometry.export_to_xml()

material=openmc.Materials()
material.cross_sections='/home/jovibuca/endf/endfb-viii.0-hdf5/cross_sections.xml'
material.export_to_xml()

setting = openmc.Settings()
setting.export_to_xml()
```
