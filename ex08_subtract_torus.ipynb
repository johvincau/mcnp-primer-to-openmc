This example showcases the use of a torus to subtract from the inner sphere.

```python
import openmc 

# Surface definitions
s1 = omc.Sphere(r=5)
s2 = omc.Sphere(r=10)
s3 = omc.XTorus(x0=4,y0=0,z0=0,a=4,b=1.5,c=1.5)
```

&: intersection, logical AND, (MCNP's ' ')
|: union, logical OR, (MCNP's ':')
~: complement, logical NOT, (MCNP's '#')

```python
# Cell definitions
c1 = omc.Cell(cell_id=1,fill=None,region=-s1 & +s3)
c2 = omc.Cell(cell_id=2,fill=None,region= -s2 | (+s1 & -s3))
c3 = omc.Cell(cell_id=3,fill=None,region=+s2)

geometry=omc.Geometry([c1,c2,c3])
geometry.export_to_xml()

material=omc.Materials()
material.cross_sections='/home/jovibuca/endf/endfb-viii.0-hdf5/cross_sections.xml'
material.export_to_xml()

setting = omc.Settings()
setting.export_to_xml()
```
