A cell consisting of overlapping spheres.

```python
import openmc

# Surface definitions
s1 = openmc.XCylinder(y0=-1,z0=-1,r=4)
s2 = openmc.XCylinder(y0=2,z0=1,r=4)
```

```python
# Cell definitions
c1 = openmc.Cell(fill=None,region=(-s2 | -s1))
c2 = openmc.Cell(fill=None,region=(+s2 & + s1))

geometry=openmc.Geometry([c1,c2])
geometry.export_to_xml()

material=openmc.Materials()
#material.cross_sections='/home/jovibuca/endf/endfb-viii.0-hdf5/cross_sections.xml'
material.export_to_xml()
setting = omc.Settings()
setting.export_to_xml()
```
