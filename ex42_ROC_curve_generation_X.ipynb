Example 42 shows a 15 MeV photon source impinging on a 10kg U238 sphere to show the generation of a receiver operating characteristic (ROC) curve using an MCNP F1 surface current tally and the `FT1 scx 1 roc` cards.

A ROC curve plots the True Positive Rate (TPR) against the False Positive Rate (FPR) for many threshold values. 

There is no way to do this natively with a base OpenMC install; one would need to develop an additional post-processing layer to compute the detector response.
