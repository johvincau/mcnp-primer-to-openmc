```python
import openmc 

# Surface definitions
s1 = openmc.XPlane(x0=-3)
s2 = openmc.XCylinder(r=2)
s3 = openmc.XPlane(x0=-1)
s4=openmc.XCylinder(r=5)
s5 = openmc.XPlane(x0=1)
s6=openmc.XCylinder(r=3.5)
s7 = openmc.XPlane(x0=3)
s8 = openmc.Sphere(r=8)
```

```python
# Cell definitions
c1 = openmc.Cell(fill=None,region=(+s1 & -s2 & -s3))
c2 = openmc.Cell(fill=None,region=(+s3 & - s4 & -s5))
c3 = openmc.Cell(fill=None,region=(+s5 &  -s6 & -s7))
c4 = openmc.Cell(fill=None,region=(~c1.region & ~c2.region & ~c3.region & -s8))
c5 = openmc.Cell(fill=None,region=(+s8))

geometry=openmc.Geometry([c1,c2,c3,c4,c5])
geometry.export_to_xml()

material=openmc.Materials()
#material.cross_sections='/home/jovibuca/endf/endfb-viii.0-hdf5/cross_sections.xml'
material.export_to_xml()

setting = openmc.Settings()
setting.export_to_xml()
```
