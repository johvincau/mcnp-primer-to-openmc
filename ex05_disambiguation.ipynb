This model demonstrates the use of a disambiguation surface (surface 7) to make cell 2 valid.

```python
import openmc

# Surface definitions
s1 = openmc.YPlane(y0=-2)
s2= openmc.YPlane(y0=2)
s3 = openmc.YCylinder(r=2)
s4=openmc.ZCylinder(r=1)
s5 = openmc.ZPlane(z0=3)
s6=openmc.Sphere(r=4)
s7 = openmc.ZPlane(z0=0) #ambiguity surface
```

```python
# Cell definitions
c1 = openmc.Cell(fill=None,region=(+s1 & -s2 & -s3))
c2 = openmc.Cell(fill=None,region=(+s3 & -s4 & -s5 & +s7))
c3 = openmc.Cell(fill=None,region=(-s6 & ~c1.region & ~c2.region))
c4 = openmc.Cell(fill=None,region=(+s6))

geometry=openmc.Geometry([c1,c2,c3,c4])
geometry.export_to_xml()

material=openmc.Materials()
#material.cross_sections='/home/jovibuca/endf/endfb-viii.0-hdf5/cross_sections.xml'
material.export_to_xml()

setting = openmc.Settings()
setting.export_to_xml()
```
